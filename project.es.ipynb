#### Ejercicio 20. Realiza el mismo gráfico para 4 poblaciones diferentes (columna level5) y colócalos en el mismo gráfico. Debes usar el subconjunto obtenido en la pregunta 14 (★★☆) 
Pista: Haz un diagrama de dispersión de cada población usando subgráficos (subplots).

```python
import matplotlib.pyplot as plt


#Definimos el eje X con los precios
XG = dfs.loc[dfs['level5']=="Getafe"]['price']
XA = dfs.loc[dfs['level5']=="Alcorcón"]['price']
XL = dfs.loc[dfs['level5']=="Leganés"]['price']
XF = dfs.loc[dfs['level5']=="Fuenlabrada"]['price']

#Definimos el eje Y para cada Localidad
YG = dfs.loc[dfs['level5']=="Getafe"]['surface']
YA = dfs.loc[dfs['level5']=="Alcorcón"]['surface']
YL = dfs.loc[dfs['level5']=="Leganés"]['surface']
YF = dfs.loc[dfs['level5']=="Fuenlabrada"]['surface']

#Grafico de Dispersión de Precio vs Area para cada localidad
plt.figure(figsize=(15,10))

plt.scatter(XG, YG, label="Getafe", color='blue')
plt.scatter(XA, YA ,label="Alcorcón", color='green')
plt.scatter(XL, YL ,label="Leganés", color='red')
plt.scatter(XF, YF,label="Fuenlabrada", color='orange')

plt.xlabel("Precios USD")
plt.ylabel("Area m2")
plt.title("Relación de Precios USD por m2")
plt.legend()

plt.show()
```
